# Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

# Simple Classification Model

## Model Build

```python
model1 = Sequential()
model1.add(Input(shape=(X.shape[1],)))
model1.add(Dense(512, activation='relu'))
model1.add(Dense(256, activation='relu'))
model1.add(Dense(512, activation='relu'))
model1.add(Dense(1, activation='sigmoid'))
```

```python
model1.summary()
```

## Model Compile

```python
model1.compile(
    optimizer=Adam(learning_rate=0.01),
    loss=BinaryCrossentropy(),
    metrics=['accuracy']
)
```

## Model Training

```python
history1 = model1.fit(
    x=X_train,
    y=y_train,
    epochs=500,
    validation_split=0.2,
    verbose=0
)
```

## evaluate the model

```python
_, train_mse = model1.evaluate(X_train, y_train, verbose=0)
_, test_mse = model1.evaluate(X_test, y_test, verbose=0)
print('Train: {}, Test: {}'.format(train_mse, test_mse))
```

## Plot Data

```python
plot_decision_regions(X_train, y_train.astype('int'), clf=model1, legend=2)
plt.xlim(-0.7,0.5)
plt.ylim(-0.8,0.8)
plt.show()
```

## Loss Plot

```python
plt.plot(history1.history['loss'])
plt.plot(history1.history['val_loss'])
plt.title('Model Loss')
plt.ylabel('Loss')
plt.xlabel('Epoch')
plt.legend(['Train', 'Test'], loc='upper left')
plt.show()
```

## Accuracy Plot

```python
plt.plot(history1.history['accuracy'])
plt.plot(history1.history['val_accuracy'])
plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend(['Train', 'Test'], loc='upper left')
plt.show()
```

# Dropout Layers Add in Classification

## Model Build

```python
model2 = Sequential()
model2.add(Input(shape=(X.shape[1],)))

model2.add(Dense(512, activation='relu'))
model2.add(Dropout(0.5))

model2.add(Dense(256, activation='relu'))
model2.add(Dropout(0.2))

model2.add(Dense(512, activation='relu'))
model2.add(Dropout(0.5))

model2.add(Dense(1, activation='sigmoid'))
```

```python
model2.summary()
```

## Model Compile

```python
model2.compile(
    optimizer=Adam(learning_rate=0.01),
    loss=BinaryCrossentropy(),
    metrics=['accuracy']
)
```

## Model Training

```python
history2 = model2.fit(
    x=X_train,
    y=y_train,
    epochs=500,
    validation_split=0.2,
    verbose=0
)
```

## evaluate the model

```python
_, train_mse = model2.evaluate(X_train, y_train, verbose=0)
_, test_mse = model2.evaluate(X_test, y_test, verbose=0)
print('Train: {}, Test: {}'.format(train_mse, test_mse))
```

## Plot Data

```python
plot_decision_regions(X_train, y_train.astype('int'), clf=model2, legend=2)
plt.xlim(-0.7,0.5)
plt.ylim(-0.8,0.8)
plt.show()
```
