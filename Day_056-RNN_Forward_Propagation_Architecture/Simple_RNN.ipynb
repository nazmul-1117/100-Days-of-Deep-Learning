# Simple_RNN

# Import Libraries

```python
from keras.models import Sequential
from keras.layers import Dense, SimpleRNN, Input
```

# Model Build

```python
model = Sequential()

model.add(Input(shape=(3, 3)))
model.add(SimpleRNN(units=3, activation='relu'))
model.add(Dense(units=1, activation='sigmoid'))
```

## Model Summary

```python
model.summary()
```
