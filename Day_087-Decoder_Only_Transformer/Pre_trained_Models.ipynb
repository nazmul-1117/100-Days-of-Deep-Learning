# Pre-trained Models

# Import Libraries

```python
from transformers import AutoModelForCausalLM, AutoTokenizer
import torch
```

# Model Load

```python
model_name = "distilgpt2"
# model_name = "Salesforce/codegen-350M-mono"
tokenizer = AutoTokenizer.from_pretrained(pretrained_model_name_or_path=model_name)
model = AutoModelForCausalLM.from_pretrained(pretrained_model_name_or_path=model_name)
```

# Device Allocations

```python
device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)
print(device)
```

# Interactive generation

```python
def generate_result(promt, max_length=25):
  inputs = tokenizer(promt, return_tensors='pt')
  output = model.generate(
                  input_ids=inputs['input_ids'],
                  attention_mask=inputs['attention_mask'],
                  max_length=max_length,
                  do_sample=True,
                  temperature=0.9,
                  pad_token_id=tokenizer.eos_token_id  # Important for GPT-2
                )

  return tokenizer.decode(output[0], skip_special_tokens=True)

while True:
  prompt = input("User: ")
  promt = prompt.lower()
  if prompt == 'exit' or not prompt or prompt == 'x':
    break

  # print(f'User: {prompt}')
  print(f'Bot: {generate_result(prompt)}\n')
```

```python
# Interactive generation
def generate_text(prompt, max_length=50):
    inputs = tokenizer(prompt, return_tensors="pt").to(device)
    outputs = model.generate(inputs.input_ids, max_length=max_length, do_sample=True, temperature=0.8)
    return tokenizer.decode(outputs[0], skip_special_tokens=True)

# Example
prompt = "Hello, how are you"
print(generate_text(prompt))
```
