# Pooling Layer in CNN

# Import Libraries

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
````

````python
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, Conv2D, Flatten, Input, MaxPooling2D, AveragePooling2D

from keras.datasets import fashion_mnist
````

# Load Dataset

````python
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
````

# Image Show

````python
plt.imshow(X_train[17])
````

# Without Pooling

````python
model1 = Sequential()

# Input Layer
model1.add(Input(shape=(28, 28, 1)))

# Convolution Layer
model1.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model1.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model1.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))

# Flatten Layer
model1.add(Flatten())

# Full Connected Layer
model1.add(Dense(128, activation='relu'))
model1.add(Dense(10, activation='softmax'))

model1.summary()
````

# Max Pooling

```python
keras.layers.MaxPooling2D(
    pool_size=(2, 2), strides=None, padding="valid", data_format=None, name=None, **kwargs
)
```

````python
model2 = Sequential()

# Input Layer
model2.add(Input(shape=(28, 28, 1)))

# Convolution Layer
model2.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model2.add(MaxPooling2D(pool_size=(2,2), strides=(2, 2), padding='same'))

model2.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model2.add(MaxPooling2D(pool_size=(2,2), strides=(2, 2), padding='same'))

model2.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model2.add(MaxPooling2D(pool_size=(2,2), strides=(2, 2), padding='same'))

# Flatten Layer
model2.add(Flatten())

# Full Connected Layer
model2.add(Dense(128, activation='relu'))
model2.add(Dense(10, activation='softmax'))
````

````python
model2.summary()
````

# Avarage Pooling

```python
keras.layers.AveragePooling2D(
    pool_size, strides=None, padding="valid", data_format=None, name=None, **kwargs
)
```

````python
model3 = Sequential()

# Input Layer
model3.add(Input(shape=(28, 28, 1)))

# Convolution Layer
model3.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model3.add(AveragePooling2D(pool_size=(2, 2), strides=(2, 2), padding='same'))

model3.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model3.add(AveragePooling2D(pool_size=(2, 2), strides=(2, 2), padding='same'))

model3.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model3.add(AveragePooling2D(pool_size=(2, 2), strides=(2, 2), padding='same'))

# Flatten Layer
model3.add(Flatten())

# Full Connected Layer
model3.add(Dense(128, activation='relu'))
model3.add(Dense(10, activation='softmax'))

model3.summary()
````
