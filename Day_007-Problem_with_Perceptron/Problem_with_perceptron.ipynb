# Import Library

```python
import  pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.linear_model import Perceptron
from sklearn.metrics import accuracy_score
```

```python
or_df = pd.DataFrame()
and_df = pd.DataFrame()
xor_df = pd.DataFrame()
xnor_df = pd.DataFrame()
```

## Or Dataset

```python
or_df['x1'] = [0,0,1,1]
or_df['x2'] = [0,1,0,1]
or_df['y'] = [0,1,1,1]
```

## and Dataset

```python
and_df['x1'] = [0,0,1,1]
and_df['x2'] = [0,1,0,1]
and_df['y'] = [0,0,0,1]
```

## xor

```python
xor_df['x1'] = [0,0,1,1]
xor_df['x2'] = [0,1,0,1]
xor_df['y'] = [0,1,1,0]
```

## xnor

```python
xnor_df['x1'] = [0,0,1,1]
xnor_df['x2'] = [0,1,0,1]
xnor_df['y'] = [1,0,0,0]
```

# Plot Diagram

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x=or_df['x1'],y=or_df['x2'],hue=or_df['y'])
plt.title('Or Dataset')
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x=and_df['x1'],y=and_df['x2'],hue=and_df['y'])
plt.title('And Dataset')
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x=xor_df['x1'],y=xor_df['x2'],hue=xor_df['y'])
plt.title('Xor Dataset')
```

```python
plt.figure(figsize=(5,3))
sns.scatterplot(x=xnor_df['x1'],y=xnor_df['x2'],hue=xnor_df['y'])
plt.title('Xnor Dataset')
```

# Model Build

```python
or_model = Perceptron()
and_model = Perceptron()
xor_model = Perceptron()
xnor_model = Perceptron()
```

```python
or_model.fit(or_df[['x1','x2']],or_df['y'])
and_model.fit(and_df[['x1','x2']],and_df['y'])
xor_model.fit(xor_df[['x1','x2']],xor_df['y'])
xnor_model.fit(xnor_df[['x1','x2']],xnor_df['y'])
```

# Plot Model

## OR

```python
m = -or_model.coef_[0][0]/or_model.coef_[0][1]
c = -or_model.intercept_[0]/or_model.coef_[0][1]
```

```python
x = np.linspace(-1, 1, 5)
y = m*x+c
plt.figure(figsize=(5,3))
sns.scatterplot(x=or_df['x1'],y=or_df['x2'],hue=or_df['y'])
plt.plot(x,y)
plt.title('Or Dataset')
```

## And

```python
m = -and_model.coef_[0][0]/and_model.coef_[0][1]
c = -and_model.intercept_[0]/and_model.coef_[0][1]
```
