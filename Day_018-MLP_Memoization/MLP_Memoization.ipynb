# Import Libraries

```python
import time
```

# Normal Approaches

```python
def fib(n: int) -> int:
  if n == 1 or n == 0:
    return n
  return fib(n-1) + fib(n-2)
```

```python
st_time = time.time()
print(f'Result: {fib(40)}')
print(f"Time taken: {time.time() - st_time:.2f}")
```

# Dynamic Programming

```python
def dp_fib(n: int, memory: dict) -> dict:
  if n in memory:
    return memory[n]
  if n <= 1:
    memory[n] = n
  else:
    memory[n] = dp_fib(n-1, memory) + dp_fib(n-2, memory)
  return memory[n]
```

```python
st_time = time.time()
print(f'Result: {dp_fib(140, {})}')
print(f"Time taken: {time.time() - st_time:.6f}")
```
