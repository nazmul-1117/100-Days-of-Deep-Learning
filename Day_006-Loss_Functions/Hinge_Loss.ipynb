# Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
from sklearn.datasets import make_classification
```

# Make Dataset

```python
X, y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=15)
```

```python
X.shape, y.shape
```

# Plot

```python
sns.scatterplot(x=X[:,0], y=X[:,1], hue=y)
```

# Perceptron

```python
class MyPerceptron:
  def __init__(self, epochs=1, learning_rate=1) -> None:
    self.w1 = None
    self.w2 = None
    self.b = None
    self.epochs = epochs
    self.learning_rate = learning_rate

  def fit(self, X, y):
    self.w1 = 1
    self.w2 = 1
    self.b = 0

    for i in range(self.epochs):
      for j in range(X.shape[0]):
        f_xi = self.w1*X[j][0] + self.w2*X[j][1] + self.b

        if y[j]*f_xi < 0:
          self.w1 = self.w1 + self.learning_rate*y[j] * X[j][0]
          self.w2 = self.w2 + self.learning_rate*y[j] * X[j][1]
          self.b = self.b + self.learning_rate*y[j]

```

```python
my_perceptron = MyPerceptron(epochs=1000, learning_rate=0.001)
my_perceptron.fit(X,y)
```

```python
my_perceptron.w1, my_perceptron.w2, my_perceptron.b
```

```python
m = -my_perceptron.w1/my_perceptron.w2
c = -my_perceptron.b/my_perceptron.w2
```

```python
m, c
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + c

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
