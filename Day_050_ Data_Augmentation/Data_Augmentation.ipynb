# Data_Augmentation

# Import Libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
%matplotlib inline
```

```python
import tensorflow as tf
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

# Image Load

```python
cat_image = image.load_img('cat_image.jpg', target_size=(256, 256))
cat_image
```

# Data Generator

```python
cat_datagen = ImageDataGenerator(
    rescale=1./255,
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='nearest'
)
```

## Convert to Numpy Array

```python
image_array = image.img_to_array(cat_image)
image_array.shape
```

## Make Batch

```python
image_batch = image_array.reshape((1,) + image_array.shape)
image_batch.shape
```

## Data Augmentation

```python
i = 0;

for batch in cat_datagen.flow(image_batch, batch_size=1, save_to_dir='aug'):
    plt.figure(i)
    imgplot = plt.imshow(image.array_to_img(batch[0]))
    i += 1

    if i % 5 == 0:
        break
```
