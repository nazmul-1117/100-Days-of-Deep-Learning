# Simple BERT Playground

# Import Libraries

```python
import torch
from transformers import BertTokenizer, BertModel
```

```python
import torch.nn.functional as F
```

# Load the pre-trained BERT model

```python
model = BertModel.from_pretrained('bert-base-uncased')
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
```

# Play with BERT!

```python
text = "hi, my name"
tokens = tokenizer(text, return_tensors='pt')
tokens
```

# Get Output

```python
with torch.no_grad():
    output = model(**tokens)
    display(output[0].shape)
```

# Get the sentence embedding

```python
cls_embedding = output[0][:, 0, :]
cls_embedding.shape
```

```python
output[0] #[:, 0, :]
```

# Try multiple sentences

```python
sentences = [
    "I love BERT!",
    "The weather is nice today.",
    "Artificial intelligence is fun."
]

inputs = tokenizer(sentences, padding=True, truncation=True, return_tensors="pt")

with torch.no_grad():
    outputs = model(**inputs)

sentence_embeddings = outputs.last_hidden_state[:, 0, :]  # CLS tokens
print(sentence_embeddings.shape)
```

# Cosine similarity (to compare meanings)

```python
sim = F.cosine_similarity(sentence_embeddings[0], sentence_embeddings[1], dim=0)
sim
```
