# Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load Data

```python
df = pd.read_csv('DailyDelhiClimateTest.csv')
df.head()
```

# Clean Column

```python
df.dropna(inplace=True)
```

```python
df.drop(columns=['meantemp', 'wind_speed', 'meanpressure'], axis=1, inplace=True)
df.head()
```

# Custom EWMA

```python
def ewma(beta, series):
    result = [series[0]]
    for n in range(1, len(series)):
        result.append(beta * series[n] + (1 - beta) * result[n - 1])
    return result
```

```python
ewma_result = ewma(0.9, df['humidity'])
```

# Pandas EWMA

```python
pd_result = df['humidity'].ewm(alpha=0.9).mean()
```

```python
display(ewma_result[:5])
display(pd_result[:5])
```
