# Non-Padding

> Python
```python
Conv2D(64, kernel_size=(3, 3), padding='valid', activation='relu')
```

````python
model1 = Sequential()

# Input Layer
model1.add(Input(shape=(28, 28, 1)))

# Convolution Layer
model1.add(Conv2D(64, kernel_size=(3, 3), padding='valid', activation='relu'))
model1.add(Conv2D(64, kernel_size=(3, 3), padding='valid', activation='relu'))
model1.add(Conv2D(64, kernel_size=(3, 3), padding='valid', activation='relu'))

# Flatten Layer
model1.add(Flatten())

# Full Connected Layer
model1.add(Dense(128, activation='relu'))
model1.add(Dense(10, activation='softmax'))
````

````python
model1.summary()
````

# Padding

> Python
```python
Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu')
```

````python
model2 = Sequential()

# Input Layer
model2.add(Input(shape=(28, 28, 1)))

# Convolution Layer
model2.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model2.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))
model2.add(Conv2D(64, kernel_size=(3, 3), padding='same', activation='relu'))

# Flatten Layer
model2.add(Flatten())

# Full Connected Layer
model2.add(Dense(128, activation='relu'))
model2.add(Dense(10, activation='softmax'))
````

````python
model2.summary()
````

# Strides

$$
O = \frac{N - F + 2P}{S} + 1
$$

$$
=> \frac{28 - 3 + 2}{2} + 1
$$

$$
=> 14
$$


````python
model3 = Sequential()

# Input Layer
model3.add(Input(shape=(28, 28, 1)))

# Convolution Layer
model3.add(Conv2D(64, kernel_size=(3, 3), strides=(2, 2), padding='same', activation='relu'))
model3.add(Conv2D(64, kernel_size=(3, 3), strides=(2, 2), padding='same', activation='relu'))
model3.add(Conv2D(64, kernel_size=(3, 3), strides=(2, 2), padding='same', activation='relu'))

# Flatten Layer
model3.add(Flatten())

# Full Connected Layer
model3.add(Dense(128, activation='relu'))
model3.add(Dense(10, activation='softmax'))
````

````python
model3.summary()
````
